# 实验1：正则表达式的基础匹配

```python
import re

s = "cat on mat; later: cats and dogs"
print(re.findall(r"\d+", s), # 匹配1或多个数字
      re.findall(r"\s+", s), # 匹配1或多个空白字符
      re.findall(r"\bcat\b", s)) # 匹配单词"cat"

print(re.findall(r"cat|dog", "cat dog bird CAT")) # | 管道符 或
print(re.findall(r"a|aa|aaa", "aaaa"), re.findall(r"aaa|aa|a", "aaaa")) # 差异优先先让最前面的匹配满，前者的话就只会匹配a，后面就是aaa a

s = "ct cat caat caaat"
for p in [r"ca*t", r"ca+t", r"ca?t", r"ca{2}t", r"ca{1,2}t"]:
    print(p, re.findall(p, s))

'''
 * = 0 或多
 + = 1 或多
 ? = 0 或 1
 {n} = 恰好 n 次
 {m,n} = m 到 n 次
'''
```

# 实验2：分组、回溯引⽤与替换

```python
import re

# 捕获组提取邮箱的 用户名 和 域名
print(re.findall(r"(\w+)@(\w+\.\w+)", "联系：alice@example.com；备选：bob@test.org")) # 这个（）就是分组，获取的数据就是元组

# 任务2：后向引用 \1 匹配连续重复的单词
print(re.findall(r"\b(\w+)\s+\1\b", "cat cat, dog cat, bird bird")) # \1 匹配连续重复的单词 \b:单词边界

# 任务3：日期 年-月-日 -> 月-日-年（保留行数）
d = "2024-05-01\n2024/05/02\n2024.05.03"
print("原始文本：", d)
print(re.sub(r"(\d{4})[-/.](\d{2})[-/.](\d{2})", r"\2-\3-\1", d)) # sub按照spel替换

# 任务4：电话统一为 (123) 456-7890
print(re.sub(r"(\d{3})[-.](\d{3})[-.](\d{4})", r"(\1) \2-\3", "联系我：123-456-7890 或 987.654.3210"))
```

# 实验3：使用 BeautifulSoup 解析 HTML

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")

# 任务1：标题文本、h1 的标签名与文本
print("title:", soup.title.text)
h1 = soup.find("h1")
print("h1:", h1.name, h1.text)

# 任务2：find_all 找出所有 class="course" 的段落和所有链接
for p in soup.find_all("p", class_="course"):
    print("p:", p.text)
for a in soup.find_all("a"):
    print("a:", a.get_text(), a["href"])

# 任务3：get_text() 提取可见文本
print("整篇文档:", repr(soup.get_text()))
print("单个段落:", repr(soup.find("p").get_text()))
print("单个链接:", repr(soup.find("a").get_text()))
```
