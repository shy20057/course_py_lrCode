## 实验 2：函数、条件、循环与递归

```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("分数必须在 0 到 100 之间")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"

for score in [95, 85, 70, 50]:
    print(score, score_level(score))
try:
    score_level(101)
except ValueError as error:
    print(error)

sum_for = 0
for number in range(1, 101):
    sum_for += number
sum_while = 0
number = 1
while number <= 100:
    sum_while += number
    number += 1
print(f"for: {sum_for}, while: {sum_while}, 相等：{sum_for == sum_while}")

def factorial_recursive(n):
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    result = 1
    for value in range(2, n + 1):
        result *= value
    return result

for n in [0, 1, 5, 10]:
    print(f"{n}!：递归={factorial_recursive(n)}，迭代={factorial_iterative(n)}")

def summary(values):
    return min(values), max(values), sum(values) / len(values)

minimum, maximum, average = summary([85, 58, 92, 76, 60, 89])
print(f"最小值={minimum}，最大值={maximum}，平均值={average:.2f}")

def sum_all(*args):
    return sum(args)

print(sum_all(), sum_all(5), sum_all(1, 2, 3, 4))
```

## 实验 3：列表、字典、元组与基础统计

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

score_by_name = dict(zip(names, scores))
for name, score in score_by_name.items():
    print(f"{name}: {score}")

for index, (name, score) in enumerate(zip(names, scores), start=1):
    print(index, name, score)

passed_names = [name for name, score in zip(names, scores) if score >= 60]
squared_scores = [score ** 2 for score in scores]
print("及格学生：", passed_names)
print("成绩平方：", squared_scores)

count = len(scores)
average_score = sum(scores) / count
highest_score = max(scores)
lowest_score = min(scores)
pass_rate = len(passed_names) / count
print(f"人数={count}，平均分={average_score:.2f}，最高分={highest_score}，最低分={lowest_score}，及格率={pass_rate:.2%}")

demo_scores = [85, 58, 92]
alias = demo_scores
copied = demo_scores.copy()
alias[0] = 100
print(f"原列表={demo_scores}，alias={alias}，副本={copied}")

student_record = ("Alice", 85)
try:
    student_record[1] = 90
except TypeError as error:
    print(f"元组元素不能重新赋值：{error}")
print(list(zip([1, 2, 3], ["a", "b"])))
```

## 实验 4：文件读写、异常处理与模块

`w` 模式会新建或覆盖原文件，`a` 模式会在原文件末尾追加内容。

```python
import pickle

print("当前工作目录：", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
students_path = os.path.join(data_dir, "students.txt")
report_path = os.path.join(data_dir, "report.txt")
pickle_path = os.path.join(data_dir, "students.pkl")

students_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(students_path, "w", encoding="utf-8") as file:
    file.write(students_text)

def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError("记录必须包含姓名和成绩两个字段")
    name = parts[0].strip()
    if not name:
        raise ValueError("姓名不能为空")
    try:
        score = int(parts[1].strip())
    except ValueError as error:
        raise ValueError("成绩必须是整数") from error
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0 到 100 之间")
    return {"name": name, "score": score}

valid_records = []
invalid_records = []
with open(students_path, encoding="utf-8") as file:
    next(file)
    for line_number, line in enumerate(file, start=2):
        try:
            valid_records.append(parse_student(line))
        except ValueError as error:
            invalid_records.append((line_number, str(error)))

print("有效记录：", valid_records)
print("无效记录：", invalid_records)
average_valid = sum(item["score"] for item in valid_records) / len(valid_records)
pass_rate_valid = sum(item["score"] >= 60 for item in valid_records) / len(valid_records)
with open(report_path, "w", encoding="utf-8") as file:
    file.write(f"有效记录数: {len(valid_records)}\n无效记录数: {len(invalid_records)}\n平均分: {average_valid:.2f}\n及格率: {pass_rate_valid:.2%}\n")
with open(report_path, "a", encoding="utf-8") as file:
    file.write("说明：w 模式覆盖已有内容；a 模式在文件末尾追加内容。\n")

with open(pickle_path, "wb") as file:
    pickle.dump(valid_records, file)
with open(pickle_path, "rb") as file:
    restored_records = pickle.load(file)
print("pickle 恢复结果与原数据相等：", restored_records == valid_records)

from student_utils import parse_student as module_parse_student
print("模块有效输入：", module_parse_student("Alice,85"))
try:
    module_parse_student("Grace,abc")
except ValueError as error:
    print("模块无效输入：", error)

try:
    with open(os.path.join(data_dir, "not_exists.txt"), encoding="utf-8") as file:
        file.read()
except FileNotFoundError:
    print("已捕获：指定文件不存在。")
```

## 实验 5：类、继承与对象复制

```python
import copy

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, research_topic={self.research_topic})"

students = [Student(item["name"], item["score"]) for item in valid_records]
for student in students:
    print(student, "及格：", student.is_passed())
graduate = GraduateStudent("Carol", 92, "Data Analysis")
print(graduate)

original = {"name": "Alice", "scores": [85, 90]}
assigned = original
assigned["scores"][0] = 100
print("直接赋值后：", original, assigned)

original = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(original)
shallow["scores"][0] = 100
print("浅拷贝后：", original, shallow)

original = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(original)
deep["scores"][0] = 100
print("深拷贝后：", original, deep)
```
