## 一、热身：定义简单函数

```python
def say_hi():
    """打印问候语。"""
    print("Hello, world!")


def goat_pad(text):
    """在字符串 text 的前后各加上 goat 并打印。"""
    print("goat" + text + "goat")


def print_n(s, n):
    """将字符串 s 打印 n 次，每次占一行。"""
    for _ in range(n):
        print(s)


# 示例调用
say_hi()
goat_pad("cheese")
print_n("Python", 3)
```

## 二、欧几里得算法

先将输入取绝对值，再不断以余数替换较小的数；当余数为 0 时，另一个数就是最大公约数。这样即使某个输入为负数，函数仍返回非负的最大公约数，符合通常的数学约定。

```python
def gcd(m, n):
    """使用欧几里得算法返回 m 和 n 的非负最大公约数。"""
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m


pairs = [(2018, 2019), (1200, 300), (5040, 60)]
for m, n in pairs:
    print(f"gcd({m}, {n}) = {gcd(m, n)}")

print(f"gcd(-1200, 300) = {gcd(-1200, 300)}")
```

## 三、近似欧拉数 $e$


```python
import math


def euler_limit(n):
    """使用 (1 + 1/n)^n 近似 e。"""
    return float((1 + 1 / n) ** n)


def euler_infinite_sum(n):
    """返回级数前 n 项（从 k=0 开始）的和。"""
    total = 0.0
    factorial = 1
    for k in range(n):
        if k > 0:
            factorial *= k
        total += 1 / factorial
    return total


def euler_approx(epsilon):
    """返回与真实 e 的误差小于 epsilon 的级数近似值。"""
    target = math.exp(1)
    total = 0.0
    term = 1.0
    k = 0
    while abs(total - target) >= epsilon:
        total += term
        k += 1
        term /= k
    return total


def print_euler_sum_table(n):
    """逐行打印 euler_infinite_sum(1) 到 euler_infinite_sum(n)。"""
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))


print(f"euler_limit(1000) = {euler_limit(1000)}")
print(f"euler_infinite_sum(4) = {euler_infinite_sum(4)}")
print(f"euler_approx(1e-6) = {euler_approx(1e-6)}")
print("前 5 项部分和：")
print_euler_sum_table(5)
```

## 四、测试整数属性

通过不断整除并检查余数，判断一个正整数是否为 2 的幂，或是否为指定底数的幂。这里将 1 视为任意正整数底数的 0 次幂。

```python
def is_power_of_2(n):
    """判断正整数 n 是否为 2 的幂。"""
    while n > 1 and n % 2 == 0:
        n //= 2
    return n == 1


def is_power(b, n):
    """判断正整数 n 是否为正整数 b 的幂。"""
    if b == 1:
        return n == 1
    while n > 1 and n % b == 0:
        n //= b
    return n == 1


for number in [1, 2, 12, 64, 100]:
    print(f"is_power_of_2({number}) = {is_power_of_2(number)}")

for base, number in [(3, 81), (3, 45), (7, 1), (1, 2)]:
    print(f"is_power({base}, {number}) = {is_power(base, number)}")
```
