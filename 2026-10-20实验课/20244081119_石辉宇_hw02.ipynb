## 一、字符串的乐趣

```python
def is_palindrome(s):
    s = s.replace(" ", "").lower()
    return s == s[::-1]


def is_abecedarian(s):
    s = s.replace(" ", "").lower()
    for i in range(len(s) - 1):
        if s[i] > s[i + 1]:
            return False
    return True


def remove_vowels(s):
    result = ""
    for ch in s:
        if ch.lower() not in "aeiou":
            result += ch
    return result


print(is_palindrome("level"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("python"))
print(is_abecedarian("adder"))
print(is_abecedarian("dog"))
print(is_abecedarian("abcd efgh xyz"))
print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("AEIOU"))
# 前两个函数先去掉空格并转为小写，这样空格和大小写就不会影响判断。
```

### 字符串函数结果说明

前 3 个布尔值依次说明 `level` 和 `T A C O cat` 在忽略空格、大小写后是回文，而 `python` 不是回文。接下来的结果表明 `adder` 和 `abcd efgh xyz` 的字母按非降序排列，`dog` 不符合要求。删除元音后，`cat` 变为 `ct`，`audio` 只剩下 `d`；全是元音的 `AEIOU` 返回空字符串，所以最后一项没有显示字符。

## 二、列表的乐趣

```python
def list_reverse(a):
    if type(a) is not list:
        raise TypeError("输入必须是列表")
    return a[::-1]


def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True


print(list_reverse([1, 2, 3]))
print(is_sorted([1, 2, 2, 5]))
print(is_sorted([1, 4, 3, 5]))
print(is_sorted("abcd"))
# `is_sorted` 每次只比较相邻的两个元素，发现顺序不对就直接返回，所以只需要遍历一次。
```

**列表部分结果：** `[1, 2, 3]` 反转后得到 `[3, 2, 1]`。含有重复元素的 `[1, 2, 2, 5]` 仍然属于升序排列；`[1, 4, 3, 5]` 因为 4 后面出现了更小的 3，所以返回 `False`。字符串同样支持索引和字符比较，因此 `abcd` 也可以传给 `is_sorted` 并得到 `True`。

## 三、作为向量的元组

```python
def is_number(x):
    return type(x) in (int, float)


def vec_scalar_mult(t, s):
    if type(t) is not tuple:
        raise TypeError("t必须是元组")
    if not is_number(s):
        raise TypeError("s必须是整数或浮点数")
    for x in t:
        if not is_number(x):
            raise TypeError("元组中的元素必须是整数或浮点数")
    return tuple(x * s for x in t)


def vec_inner_product(t1, t2):
    if type(t1) is not tuple or type(t2) is not tuple:
        raise TypeError("两个输入都必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个元组长度不一致")
    for x in t1 + t2:
        if not is_number(x):
            raise TypeError("元组中的元素必须是整数或浮点数")
    total = 0
    for i in range(len(t1)):
        total += t1[i] * t2[i]
    return float(total)


def check_valid_mx(mx):
    if type(mx) is not tuple:
        return False
    if len(mx) == 0:
        return True
    for row in mx:
        if type(row) is not tuple:
            return False
        if len(row) != len(mx[0]):
            return False
        for x in row:
            if not is_number(x):
                return False
    return True


def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise TypeError("输入的矩阵不合法")
    if type(v) is not tuple:
        raise TypeError("向量必须是元组")
    for x in v:
        if not is_number(x):
            raise TypeError("向量元素必须是整数或浮点数")
    column_num = len(mx[0]) if len(mx) > 0 else 0
    if column_num != len(v):
        raise ValueError("矩阵列数和向量长度不一致")
    answer = []
    for row in mx:
        answer.append(vec_inner_product(row, v))
    return tuple(answer)


print(vec_scalar_mult((1, 2, 3), 2))
print(vec_inner_product((1, 2, 3), (4, 5, 6)))
mx = ((1, 2, 3), (4, 5, 6))
print(check_valid_mx(mx))
print(check_valid_mx(((1, 2), (3,))))
print(mx_vec_mult(mx, (1, 0, -1)))
# 矩阵乘向量时，把矩阵的每一行分别和向量做内积，得到的结果再组成一个元组。
```

### 元组向量和矩阵结果说明

向量 `(1, 2, 3)` 乘以标量 2 得到 `(2, 4, 6)`；它与 `(4, 5, 6)` 的内积为 $1\times4+2\times5+3\times6=32$。规则矩阵通过检查得到 `True`，行长度不一致的矩阵得到 `False`。矩阵乘向量时，两行分别与 `(1, 0, -1)` 做内积，结果都是 -2，所以最终结果为 `(-2.0, -2.0)`。

## 四、更多向量的乐趣

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if type(dimension) is not int:
            raise TypeError("维度必须是整数")
        if dimension < 0:
            raise ValueError("维度不能是负数")
        if elements is None:
            elements = [0] * dimension
        if type(elements) not in (list, tuple):
            raise TypeError("元素必须使用列表或元组")
        if len(elements) != dimension:
            raise ValueError("维度和元素个数不一致")
        for x in elements:
            if not is_number(x):
                raise TypeError("每个元素都必须是整数或浮点数")
        self.dimension = dimension
        self.elements = tuple(elements)

    def check_other(self, other):
        if not isinstance(other, Vector):
            raise TypeError("另一个对象必须是Vector")
        if self.dimension != other.dimension:
            raise ValueError("两个向量维度不一致")

    def __eq__(self, other):
        self.check_other(other)
        return self.elements == other.elements

    def __lt__(self, other):
        self.check_other(other)
        return self.elements < other.elements

    def __le__(self, other):
        self.check_other(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self.check_other(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self.check_other(other)
        return self.elements >= other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot的参数必须是Vector")
        if self.dimension != other.dimension:
            raise ValueError("两个向量维度不一致")
        total = 0
        for i in range(self.dimension):
            total += self.elements[i] * other.elements[i]
        return float(total)

    def __mul__(self, other):
        if is_number(other):
            result = [x * other for x in self.elements]
            return Vector(self.dimension, result)
        if isinstance(other, Vector):
            if self.dimension != other.dimension:
                raise ValueError("两个向量维度不一致")
            result = []
            for i in range(self.dimension):
                result.append(self.elements[i] * other.elements[i])
            return Vector(self.dimension, result)
        raise TypeError("只能和标量或Vector相乘")

    def __rmul__(self, other):
        return self * other


v = Vector(2, (2, 4))
w = Vector(2, (2, 5))
print("v的元素：", v.elements)
print("w的元素：", w.elements)
print("v==w：", v == w)
print("v<w：", v < w)
print("v<=w：", v <= w)
print("w>v：", w > v)
print("v和w的内积：", v.dot(w))
print("2*v：", (2 * v).elements)
print("v*2：", (v * 2).elements)
print("v*w：", (v * w).elements)
print("三维零向量：", Vector(3).elements)
# 比较时直接利用元组本身的字典序。标量乘法和两个向量逐元素相乘都会返回新的 `Vector`，原来的向量不会改变。
```

### Vector 类结果说明

`v=(2,4)` 与 `w=(2,5)` 的坐标不同，因此不相等；按照元组的字典序，先比较第一个坐标，相同时再比较第二个坐标，所以 `v<w` 和 `v<=w` 都为 `True`。两者内积为 $2\times2+4\times5=24$。左乘和右乘标量 2 都得到 `(4, 8)`，说明 `__mul__` 与 `__rmul__` 的结果一致；两个向量相乘执行对应元素相乘，结果为 `(4, 20)`。只给出维度 3 时则创建了 `(0, 0, 0)`。
